## Basic Mean

```python
import pandas as pd
numbers  = pd.Series([10, 20, 30, 40, 50])
df = numbers.mean()
print(df)
```

## Student Marks

```python
marks = pd.Series([75, 80, 65, 90, 85])
df = marks.mean()
print(df)
```

## Ages

```python
ages = pd.Series([18, 20, 21, 19, 22, 20])
df = ages.mean()
print(df)
```

## Monthly Sales

```python
sales = pd.Series([1200, 1500, 1800, 1300, 2000, 1700])
average_sale = sales.mean()
print(average_sale)
```

## Temperature

```python
Temperature = pd.Series([25, 28, 30, 27, 29, 31, 26])
average = Temperature.mean()
print(average)
```

## Mean with Labels

```python
marks  = [80,75,90,85,70]
student = ["Ali", "Ahmed", "Wahid", "Usman", "Hamza"]
df = pd.Series(marks , index= student)
print(df)
average = df.mean()
print(average)
```

## Mean of a DataFrame Column

```python
import pandas as pd

data = {
    "Name": ["Ali", "Ahmed", "Wahid", "Usman", "Hamza"],
    "Marks": [80, 75, 90, 85, 70]
}

df = pd.DataFrame(data)

print(df)

average = df["Marks"].mean()

print("Average Marks:", average)
```

## Mean of Multiple Columns

```python
data = {
    'Maths ': [55,66,77,88,99],
    'English' : [11,22,33,44,55],
    'Computer' : [56,67,78,89,90]
}
df = pd.DataFrame(data)
print(df)
Multiple = df.mean()
print(Multiple)
```

## Mean with Missing Values

```python
numb = pd.Series([11,22,33,44,55,66])
Missing = numb.isnull()
find = Missing.mean()
print(find)
```
