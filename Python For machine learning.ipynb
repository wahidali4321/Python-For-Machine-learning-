## Basic Mean

```python
import pandas as pd
numbers  = pd.Series([10, 20, 30, 40, 50])
df = numbers.mean()
print(df)
```

## Mean with Labels

```python
marks  = [80,75,90,85,70]
student = ["Ali", "Ahmed", "Wahid", "Usman", "Hamza"]
df = pd.Series(marks , index= student)
print(df)
average = df.mean()
print(average)
```

## Mean of a DataFrame Column

```python
import pandas as pd

data = {
    "Name": ["Ali", "Ahmed", "Wahid", "Usman", "Hamza"],
    "Marks": [80, 75, 90, 85, 70]
}

df = pd.DataFrame(data)

print(df)

average = df["Marks"].mean()

print("Average Marks:", average)
```

## Mean of Multiple Columns

```python
data = {
    'Maths ': [55,66,77,88,99],
    'English' : [11,22,33,44,55],
    'Computer' : [56,67,78,89,90]
}
df = pd.DataFrame(data)
print(df)
Multiple = df.mean()
print(Multiple)
```

## Mean with Missing Values

```python
numb = pd.Series([11,22,33,44,55,66])
Missing = numb.isnull()
find = Missing.mean()
print(find)
```

Real-World Dataset ⭐

```python
import pandas as pd

df = pd.read_csv("data.csv")

print(df)

age_mean = df["Age"].mean()
salary_mean = df["Salary"].mean()
rating_mean = df["Rating"].mean()

print("Average Age:", age_mean)
print("Average Salary:", salary_mean)
print("Average Rating:", rating_mean)
```

## Median of Multiple Columns

```python
import pandas as pd

data = {
    "Maths": [55, 66, 77, 88, 99],
    "English": [60, 70, 80, 90, 100],
    "Computer": [50, 65, 75, 85, 95]
}

df = pd.DataFrame(data)

print(df)

median = df.median()

print("\nMedian of each subject:")
print(median)
```

## Median by Group

```python
import pandas as pd

data = {
    "Department": ["IT", "IT", "IT", "HR", "HR", "HR"],
    "Salary": [50000, 60000, 70000, 45000, 50000, 55000]
}

df = pd.DataFrame(data)

print(df)

median_salary = df.groupby("Department")["Salary"].median()

print("\nMedian Salary:")
print(median_salary)
```
